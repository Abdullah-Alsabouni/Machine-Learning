```python
import pandas as pd
df = pd.read_csv("heart.csv")
df
```

```python
import numpy as np
import scipy.stats as stats

numeric_cols = df.select_dtypes(include=[float, int]).columns # Select numaric variables

df_zscore = df.copy()
df_zscore[numeric_cols] = df[numeric_cols].apply(stats.zscore)

df_zscore.head()
```

```python
# Remove outlires
df_no_outliers = df[(df_zscore[numeric_cols] < 3).all(axis=1) & (df_zscore[numeric_cols] > -3).all(axis=1)]
df_no_outliers
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df_no_outliers['Sex'] = le.fit_transform(df_no_outliers['Sex'])
df_no_outliers['ChestPainType'] = le.fit_transform(df_no_outliers['ChestPainType'])
df_no_outliers['RestingECG'] = le.fit_transform(df_no_outliers['RestingECG'])
df_no_outliers['ExerciseAngina'] = le.fit_transform(df_no_outliers['ExerciseAngina'])
df_no_outliers['ST_Slope'] = le.fit_transform(df_no_outliers['ST_Slope'])
df_no_outliers
```

```python
X = df_no_outliers.drop("HeartDisease", axis='columns')
y = df_no_outliers.HeartDisease
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled[:3]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=20)
```

```python
from sklearn.svm import SVC
model = SVC()
```

```python
from sklearn.svm import SVC
from sklearn.model_selection import cross_val_score

scores = cross_val_score(SVC(), X, y, cv=5)
scores.mean()
```

```python
model.fit(X_train, y_train)
```

```python
model.score(X_test, y_test)
```

```python

from sklearn.ensemble import BaggingClassifier

bag_model = BaggingClassifier(
    estimator = SVC(), 
    n_estimators=100, 
    max_samples=0.8, 
    oob_score=True,
    random_state=0
)
bag_model.fit(X_train, y_train)
bag_model.oob_score_
```

```python
bag_model.score(X_test, y_test)
```

```python
from sklearn.model_selection import cross_val_score
from sklearn.tree import DecisionTreeClassifier

scores = cross_val_score(DecisionTreeClassifier(), X, y, cv=5)
scores
```

```python
scores.mean()
```

```python
from sklearn.ensemble import BaggingClassifier

bag_model = BaggingClassifier(
    estimator=DecisionTreeClassifier(), 
    n_estimators=100, 
    max_samples=0.9, 
    oob_score=True,
    random_state=0
)
bag_model.fit(X_train, y_train)
bag_model.oob_score_
```

```python
bag_model.score(X_test, y_test)
```

```python
bag_model = BaggingClassifier(
    estimator=DecisionTreeClassifier(), 
    n_estimators=100, 
    max_samples=0.9, 
    oob_score=True,
    random_state=0
)
scores = cross_val_score(bag_model, X, y, cv=5)
scores
```

```python
scores.mean()
```

```python
from sklearn.ensemble import RandomForestClassifier

scores = cross_val_score(RandomForestClassifier(), X, y, cv=5)
scores.mean()
```
